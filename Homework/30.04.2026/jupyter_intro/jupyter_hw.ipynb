```python
# QUESTION7

str_name = "Giora_Glovatsky Giora Glovatsky"

list_name = str_name.split()
# Printing all the values in the even indexes
for i in range(len(list_name)):
    if i % 2 == 0:
        print(list_name[i])
# Printing all the values in the odd indexes
for i in range(len(list_name)):
    if i % 2 == 1:
        print(list_name[i])
```

```python
# QUESTION8

a = "#####"
b = "*****"
c = "Giora Glovatsky"

# The concatenated name
concatenate = a + c + b

# Calculating the first name and last name len
last_len = concatenate.find(b[0]) - concatenate.find(" ") - 1
name_len = concatenate.find(" ") - concatenate.rfind(a[0]) - 1

# Printing the length of the first name + last name
print(f"First name length: {name_len}, last name length: {last_len}.")
```
